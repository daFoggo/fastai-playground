## 1. Self-supervised learning là gì?
- Huấn luyện một model mà không cần phải gán nhãn dữ liệu trước, thay vào đó sử dụng nhãn từ chính dữ liệu. Bằng cách che giấu 1 phần giữ liệu và bắt model đoán lại phần bị giấu -> phần bị giấu chính là nhãn. Self-supervised learning giúp mô hình học được lượng dữ liệu khổng lồ mà không cần tốn công label của con người.

## 2. Language model là gì?
- Language model là mô hình được huấn luyện nhằm đoán ra cụm từ kế tiếp trong văn bản.

## 3. Vì sao LM là self-supervised?
- Vì nhãn đã được lấy từ văn bản được huấn luyện như giải thích câu 1.

## 4. Self-supervised model thường dùng để làm gì?
- Sử dụng để tiền huấn luyện một model phục vụ transfer learning.

## 5. Vì sao fine-tune LM trước khi làm classifier?
-  Giúp model hiểu được và thích ứng với văn phong của văn bản, giúp nó dự đoán tốt hơn các từ và cho ra kết quả chính xác hơn.
- Ví dụ muốn huấn luyện 1 model phân loại review phim, thay vì huấn luyện classification trực tiếp lên model, ta cho model làm quen trước với văn phong của review, sau đó mới đem đi dùng cho classification.

## 6. 3 bước tạo classifier SOTA?
- LM model pretrained -> Fine-tune LM với văn bản -> Fine-tune classifier

## 7. 50,000 review không nhãn giúp gì?
- Sử dụng làm training data set để fine-tune LM vì self-supervised ko cần gán nhãn trước

## 8. 3 bước chuẩn bị data cho LM?
- Tokenization: Phân các đoạn văn bản thành các từ / nhóm chữ cái / chữ cái bằng các phương pháp như khoảng trắng, dấu câu, từ phổ biến, gán kí tự đặc biệt,..
- Numerization: Chuyển đổi từ các từ thành các con số, các con số ở đây là index của hàng dữ liệu tương ứng của từ trong embeding vector array của model. -> Số bao nhiêu tùy thuộc model.
- LLMDataLoader - Batches: Chuẩn bị loader để chạy huấn luyện dữ liệu, sử dụng kĩ thuật chia patch để tối ưu performance.

## 9. Tokenization là gì, vì sao cần?
- Chuyển các đoạn văn bản thành các từ / nhóm chữ cái / chữ cái bằng các phương pháp như khoảng trắng, dấu câu, từ phổ biến, gán kí tự đặc biệt,..
- Cần có tokenization vì model cần input dạng số chứ không xử lí trực tiếp cả cụm văn bản.

## 10. 3 cách tokenization?
- Word-based: Chia một câu thành các cụm từ dựa trên khoảng trắng, dấu câu,...
- Subword-based: Chia một từ thành nhiều nhóm chữ cái nhỏ hơn nữa nếu từ quá dài và có các cụm chữ cái phổ biến như "sion", ...
- Character-based: Chia cả câu thành từng chữ cái

## 11. xxbos là gì?
- Một cụm kí tự đặc biệt nhằm đánh dấu cụm từ ở sau nó là phần bắt đầu 1 văn bản mới.
- Đây là rule prefix của fastai tokenizer


## 12. 4 rule fastai áp dụng khi tokenize
- fix_html: sửa ký tự HTML đặc biệt thành dạng đọc được
- replace_rep: ký tự lặp ≥3 lần → xxrep + số lần + ký tự
- replace_wrep: từ lặp ≥3 lần → xxwrep + số lần + từ
- spec_add_spaces: thêm khoảng trắng quanh / và #


## 13.  Vì sao ký tự lặp được thay bằng token + số lần lặp?
- Giúp model có cách hiểu chung về việc lặp từ trong văn bản, thay vì phải huấn luyện để nó tự nhận diện được việc lặp từ, thì chỉ bằng cách đọc prefix nó sẽ biết luôn.

## 14. Numericalization là gì?
- Chuyển đổi từ các từ thành các con số, các con số ở đây là index của hàng dữ liệu tương ứng của từ trong embeding vector array của model. -> Số bao nhiêu tùy thuộc model.

## 15. Vì sao có từ bị thay bằng xxunk?
- xxunk = từ không biết. Do numericaization chỉ chạy trên 60k từ phổ biến hoặc nó có số lần lặp quá ít, nên các từ ngoài list đó sẽ ko đóng góp nhiều ý nghĩa trong quá trình huấn luyện -> loại bỏ để giúp giảm kích cỡ embedding và tối ưu hiệu năng, tài nguyên.

## 16. Batch size 64: hàng 2 của batch 1 chứa gì? Hàng 1 của batch 2 chứa gì?
- Hàng 2 của batch 1 chứa phần đầu của nhóm token thứ 2.
- Hàng 1 của batch 2 chứa phần hai của nhóm  token thứ 1.

## 17. Vì sao classification cần padding, LM thì không?
- Vì với classifier, mỗi document là 1 sample riêng biệt. Khi sử dụng Pytorch, dataloader cần tensor có 1 shape cố định -> Người ta sử dụng kĩ thuật padding - thêm token fake (model sẽ bỏ qua), để giúp các tensor có độ dài như nhau.

## 18. Embedding matrix trong NLP chứa gì? Shape thế nào?
- Embedding matrix trong NLP là array đa chiều với mỗi hàng hàng sẽ biểu diễn 1 token - word dưới dạng con số.
- Shape: (vocab_size, embedding_dim)

## 19. Perplexity là gì?
- Một metric phổ biến để đo chất lượng LM.
- Giá trị = mũ của hàm loss

## 20. Vì sao phải truyền vocab của LM cho classifier DataBlock?
- Để chắc chắn rằng ánh xạ token -> index vocab y hệt nhau

## 21. Gradual unfreezing là gì?
- Gradual unfreezing là cách fine-tune model theo từng đợt: ban đầu chỉ cho phép layer cuối (Head - phần mới, ngẫu nhiên) học, các layer trước đó (đã pretrained tốt) bị khóa cứng. Sau vài epoch, mở khóa thêm 1 nhóm layer gần đó, rồi tiếp tục mở dần lùi về phía đầu vào, cho đến khi toàn bộ model được unfreeze.
- Làm vậy để tránh catastrophic forgetting: nếu mở khóa hết ngay từ đầu, gradient lỗi từ Head (còn ngẫu nhiên, đoán bậy) sẽ lan ngược và phá hỏng kiến thức ngôn ngữ tốt mà các layer đầu đã học được từ pretraining.

## 22. Vì sao sinh văn bản (generation) luôn "đi trước" khả năng phát hiện văn bản máy tạo?
- Sinh văn bản luôn đi trước phát hiện vì đây là một vòng lặp "mèo vờn chuột": mỗi khi bộ phát hiện (discriminator) học được cách nhận ra văn bản giả, chính thông tin đó lại được dùng để cải tiến bộ sinh (generator) sao cho qua mặt được dấu hiệu vừa bị phát hiện. Bộ sinh luôn có lợi thế vì nó "sửa sau" - biết chính xác lỗi nào vừa bị lộ để né tiếp, còn bộ phát hiện không thể biết trước lỗi tiếp theo sẽ là gì.
